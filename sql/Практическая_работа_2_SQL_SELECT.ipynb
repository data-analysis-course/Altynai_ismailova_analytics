### Задание 1. Операторы сравнения

Выведите название товара, цену и количество на складе.

Найдите товары, цена которых **больше 2 000**, но **не равна 2 500**.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT product_name, price, stock_quantity
FROM products
WHERE price > 2000 AND price <> 2500;
```

### Задание 2. Логический оператор AND

Выведите имя покупателя, город и email.

Найдите покупателей, которые проживают в **Bishkek** и чей email заканчивается на `gmail.com`.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT full_name, city, email
FROM customers
WHERE city = 'Bishkek' AND email LIKE '%gmail.com';
```

### Задание 3. Логический оператор OR

Выведите номер заказа, статус и сумму заказа.

Найдите заказы со статусом `completed` или `cancelled`.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT order_id, status, total_amount
FROM orders
WHERE status = 'completed' OR status = 'cancelled';
```

### Задание 4. Оператор BETWEEN

Выведите название товара и цену.

Найдите товары стоимостью **от 1 000 до 10 000 включительно**.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT product_name, price
FROM products
WHERE price BETWEEN 1000 AND 10000;
```

### Задание 5. Оператор IN

Выведите имя покупателя и город.

Найдите покупателей, которые проживают в **Bishkek, Osh или Naryn**.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT full_name, city
FROM customers
WHERE city IN ('Bishkek', 'Osh', 'Naryn');
```

### Задание 6. Оператор NOT IN

Выведите номер заказа, статус и сумму.

Найдите заказы, статус которых не равен `completed` и `cancelled`.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT order_id, status, total_amount
FROM orders
WHERE status NOT IN ('completed', 'cancelled');
```

### Задание 7. Оператор LIKE

Выведите название товара и цену.

Найдите товары, в названии которых встречается `SQL` или `Book`.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT product_name, price
FROM products
WHERE product_name ILIKE '%SQL%' OR product_name ILIKE '%Book%';
```

### Задание 8. Оператор DISTINCT

Выведите список уникальных городов покупателей.

Повторяющиеся города не должны отображаться.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT DISTINCT city
FROM customers;
```

### Задание 9. ORDER BY и LIMIT

Выведите название товара, цену и количество на складе.

Отсортируйте товары по цене от самой дорогой к самой дешёвой и покажите только первые **5 записей**.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT product_name, price, stock_quantity
FROM products
ORDER BY price DESC
LIMIT 5;
```

### Задание 10. Комплексное задание

Выведите номер заказа, сумму и статус.

Найдите заказы, которые соответствуют одному из двух условий:

- сумма от **5 000 до 20 000 включительно**, а статус — `completed` или `pending`;
- **либо** сумма меньше **4 000**, а статус — `processing`.

Отсортируйте результат по сумме от большей к меньшей.

```python
%%sql
-- Напишите SQL-запрос здесь
SELECT order_id, total_amount, status
FROM orders
WHERE (total_amount BETWEEN 5000 AND 20000 AND status IN ('completed', 'pending'))
   OR (total_amount < 4000 AND status = 'processing')
ORDER BY total_amount DESC;
```
