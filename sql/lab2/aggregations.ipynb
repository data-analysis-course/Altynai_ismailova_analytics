Задание 5. Анализ успешных платежей
Бизнес-ситуация: финансовому менеджеру необходимо оценить успешные платежи.
Таблица: payments.
Необходимо получить:
•	количество успешных платежей;
•	общую сумму успешных платежей;
•	среднюю сумму успешного платежа;
•	максимальную сумму успешного платежа.
Условие: учитывать только payment_status = 'Paid'.
Использовать: COUNT(), SUM(), AVG(), MAX(), WHERE.


```python
select * from payments
```

```python
SELECT
    COUNT(*)    AS paid_count,
    SUM(amount) AS total_amount,
    AVG(amount) AS avg_amount,
    MAX(amount) AS max_amount
FROM payments
WHERE payment_status = 'Paid';
```


Задание 7. Анализ выручки
Бизнес-ситуация: руководство хочет узнать общую стоимость всех товарных позиций, проданных через интернет-магазин.
Таблица: order_items.
Необходимо получить:
•	общую выручку;
•	среднюю стоимость товарной позиции;
•	минимальную стоимость товарной позиции;
•	максимальную стоимость товарной позиции.
Использовать: SUM(), AVG(), MIN(), MAX() и вычисление quantity * unit_price.


```python
select sum(unit_price*quantity) as revenue, avg(unit_price), min(unit_price), max(unit_price) from order_items
```
