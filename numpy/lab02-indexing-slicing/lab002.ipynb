```python
import numpy as np

order_id = np.arange(2001, 2013)

region = np.array([ "Bishkek", "Osh", "Bishkek", "Karakol", "Osh", "Tokmok", "Bishkek", "Osh", "Karakol", "Tokmok", "Bishkek", "Osh" ])

quantity = np.array([2, 1, 5, 3, 2, 8, 1, 4, 2, 3, 6, 1])

unit_price = np.array([ 1500, 20000, 800, 5000, 1200, 700, 25000, 3000, 4500, 2000, 900, 18000 ])

discount = np.array([ 0.05, 0.10, 0.00, 0.15, 0.05, 0.20, 0.30, 0.10, 0.00, 0.05, 0.50, 0.10 ])

unit_cost = np.array([ 1000, 16000, 600, 4000, 900, 500, 18000, 2200, 3500, 1500, 800, 14000 ])

revenue = quantity * unit_price * (1 - discount) 
profit = revenue - quantity * unit_cost 

```

## Задание 3. Поиск убыточных заказов — 15 баллов

Найдите заказы с отрицательной валовой прибылью.

Рассчитайте долю таких заказов среди всех заказов. Объясните, почему высокая выручка сама по себе не гарантирует высокую прибыль.


```python
loss_mask = profit < 0                       # отрицательная валовая прибыль
loss_share = loss_mask.sum() / len(profit)   # доля убыточных заказов

print("убыточные заказы:", order_id[loss_mask])
print("прибыль по ним:  ", profit[loss_mask])
print("выручка по ним:  ", revenue[loss_mask])
print(f"доля убыточных: {loss_share:.1%}  ({loss_mask.sum()} из {len(profit)})")
print("суммарный убыток:", profit[loss_mask].sum())
```

**Вывод.** Убыточны 2 заказа из 12 (≈16,7%): 2007 и 2011 (оба из Бишкека). Заказ 2007 имеет одну из самых высоких выручек (17 500), но убыток -500: после скидки 30% цена за единицу (17 500) ниже себестоимости (18 000). Выручка не учитывает затраты: прибыль = выручка - себестоимость, поэтому по выручке нельзя судить о доходности.

## Задание 4. Анализ скидок — 10 баллов

Сравните заказы с низкими и высокими скидками. Определите, какие заказы сочетают большую скидку с низкой или отрицательной прибылью.


Не делайте вывод о причинном влиянии скидки только на основании такого сравнения.

```python
avg_disc = discount.mean()
low_mask = discount < avg_disc
high_mask = discount >= avg_disc

for label, m in [("низкие скидки", low_mask), ("высокие скидки", high_mask)]:
    print(f"{label}: заказов={m.sum()}, ср.скидка={discount[m].mean():.2f}, "
          f"ср.выручка={revenue[m].mean():.0f}, ср.прибыль={profit[m].mean():.0f}, "
          f"маржа={profit[m].sum()/revenue[m].sum():.1%}")

# большая скидка + низкая прибыль (ниже медианы прибыли)
bad_mask = high_mask & (profit < np.median(profit))
print("\nбольшая скидка и низкая прибыль:", order_id[bad_mask], profit[bad_mask])
```

**Вывод.** Группа с высокими скидками (порог - средняя скидка ≈ 0,13) имеет заметно меньшую маржу, а заказы 2007 и 2011 убыточны. Однако выборка из 12 заказов мала, а скидки могли даваться на товары с низкой маржой или крупным клиентам - поэтому делать вывод, что скидка *вызывает* падение прибыли, нельзя: это лишь связь, требующая проверки.

## Задание 5. Подозрительные наблюдения

```python
rule_A = (discount >= 0.30) & (profit <= 0)                   # глубокая скидка + нулевая/отрицательная прибыль
rule_B = (unit_price > 2 * np.median(unit_price)) & (quantity <= 2)   # высокая цена при малом количестве
suspicious = rule_A | rule_B

print("медианная цена:", np.median(unit_price))
print("A (скидка>=30% и прибыль<=0):", order_id[rule_A])
print("B (цена > 2*медиана, qty<=2):", order_id[rule_B])
print("подозрительные (A или B):    ", order_id[suspicious])
print("прибыль:", profit[suspicious], "скидка:", discount[suspicious])
```

**Вывод.**
- **2011** (скидка 50%, прибыль -2100) и **2007** (скидка 30%, убыток -500, самая высокая цена 25 000) - требуют обязательной проверки: продажа ниже себестоимости, возможны ошибка цены/скидки, нарушение полномочий менеджера.
- **2002** (цена 20 000) и **2012** (цена 18 000), по 1 шт., - высокая цена при малом количестве, но маржа положительная: обычные штучные дорогие товары, особых вопросов нет.
- Остальные заказы с небольшими скидками и положительной прибылью объясняются обычной торговлей.

## Задание 7. Влияние аномалий

```python
ok = ~suspicious
print("средняя выручка, все заказы:       ", revenue.mean())
print("средняя выручка без подозрительных:", revenue[ok].mean())
print("изменение:", revenue[ok].mean() - revenue.mean())

# чувствительность к правилу
rules = {"A": rule_A, "B": rule_B, "A|B": rule_A | rule_B, "убыточные": profit < 0}
for name, r in rules.items():
    print(f"без правила {name:10s}: исключено {r.sum()}, ср.выручка {revenue[~r].mean():.0f}")
```

**Вывод.** Результат зависит от правила: правило A исключает убыточные заказы со скидкой, правило B (2002, 2007, 2012) - дорогие штучные, и средняя выручка сдвигается по-разному (все заказы: 8855; без A: 8606; без A|B: 6482). Заказы 2002/2012 имеют высокую выручку, но легитимны, поэтому автоматически удалять их нельзя - это исказило бы оценку типичного заказа. Исключать стоит только то, что подтверждено проверкой как ошибка.

## Дополнительное задание

```python
# Правило 1: бизнес-логика - прибыль < 0 или скидка >= 30%
r1 = (profit < 0) | (discount >= 0.30)
# Правило 2: статистическое - выручка вне mean +- 1 std
z = (revenue - revenue.mean()) / revenue.std()
r2 = np.abs(z) > 1

print("Правило 1:", order_id[r1])
print("Правило 2:", order_id[r2])
print("совпадают:", order_id[r1 & r2])
print("только 1: ", order_id[r1 & ~r2])
print("только 2: ", order_id[r2 & ~r1])
```

**Вывод.** Правило 1 указывает прямо на финансовый риск (убыток/глубокая скидка) и подходит для первичной проверки. Правило 1 даёт 2 заказа (2007, 2011), правило 2 - 6 (2001, 2002, 2005, 2007, 2011, 2012). Правило 2 реагирует на размер заказа, а не на убыточность и помечает прибыльные заказы (2001, 2002, 2005, 2012) - это ложные срабатывания; для первичной проверки подходит правило 1.
