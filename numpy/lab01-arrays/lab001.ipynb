Найдите:

Сформируйте отдельную булеву маску для каждого типа проблемы.

```python
def find_0_neg_vals(arr: np.array):
    
    return arr<=0
def skipped_vals(arr):
    return np.isnan(arr)
    
def unacceptable_discounts(arr):
    return ( 0>arr)|( arr>1)
def proverka_kachestva(quantity, price, discount):
    return  ~(find_0_neg_vals(quantity) | skipped_vals(price) | unacceptable_discounts(discount))

print(proverka_kachestva(quantity, unit_price,discount))
```
