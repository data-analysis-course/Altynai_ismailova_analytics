# Блок 3 — Подведение итогов по продажам

```python
import numpy as np

# Данные те же, что и в блоке 1: seed фиксированный, поэтому числа совпадают
rng = np.random.default_rng(seed=2026)

# Продажи за неделю 1: 7 дней, 3 категории товаров
week1 = rng.integers(low=500, high=2000, size=(7, 3))

# Продажи за неделю 2: 7 дней, 3 категории товаров
week2 = rng.integers(low=500, high=2000, size=(7, 3))

# Категории товаров (столбцы)
categories = np.array(["Electronics", "Clothing", "Home"])
```

## Задание 3.1 — Агрегация (на данных `week1`)

Про оси: `axis=0` — «сверху вниз», то есть схлопываем строки и получаем результат по каждому столбцу (категории).
`axis=1` — «слева направо», схлопываем столбцы и получаем результат по каждой строке (дню).

```python
# 1. Общая выручка за неделю
total_revenue = np.sum(week1)
print("Общая выручка за неделю:", total_revenue)

# 2. Средняя дневная выручка = общая выручка / число дней (по формуле из задания)
number_of_days = week1.shape[0]
average_daily_revenue = total_revenue / number_of_days
print("Средняя выручка за день:", round(average_daily_revenue, 2))

# 3. Максимум и минимум среди всех ячеек таблицы
print("Максимальное значение:", np.max(week1))
print("Минимальное значение:", np.min(week1))

# 4. Выручка по каждой категории (axis=0 — суммируем по дням)
revenue_by_category = np.sum(week1, axis=0)
for category_name, category_revenue in zip(categories, revenue_by_category):
    print(f"{category_name}: {category_revenue}")

# 5. Средняя выручка по каждому дню (axis=1 — усредняем по трём категориям)
average_revenue_per_day = np.mean(week1, axis=1)
print("Среднее по категориям для каждого дня:", np.round(average_revenue_per_day, 1))
```

Заметка: в пункте 2 мы делим на 7 дней общую выручку, поэтому получаем средний итог за *день*.
В пункте 5 мы усредняем *внутри* дня по трём категориям, так что это средний чек категории в этот день.
Числа отличаются ровно в три раза — это нормально, это разные вещи.

## Задание 3.2 — Накопленная выручка

$$
\text{Cumulative Sales}(n) = \text{Sales}(1) + \text{Sales}(2) + \dots + \text{Sales}(n)
$$

```python
# 1. Выручка за каждый день — сумма по трём категориям
daily_revenue = np.sum(week1, axis=1)
print("Выручка по дням:", daily_revenue)

# 2. Накопленная выручка: каждый элемент = сумма всех дней до него включительно
cumulative_revenue = np.cumsum(daily_revenue)

# 3. Выводим
print("Накопленная выручка:", cumulative_revenue)

# Проверка: последнее накопленное значение должно равняться общей выручке
print("Последнее значение совпадает с общей выручкой:", cumulative_revenue[-1] == total_revenue)

# 5. Когда накопленная выручка впервые превысила половину недельной
half_of_week_revenue = total_revenue / 2
exceeded_half = cumulative_revenue > half_of_week_revenue
# argmax на булевом массиве возвращает индекс первого True; прибавляем 1, чтобы получить номер дня
day_when_half_passed = np.argmax(exceeded_half) + 1
print("Половина недельной выручки:", half_of_week_revenue)
print("Половину превысили в день №", day_when_half_passed)
```

**Что показывает последнее значение?**
Накопленная выручка на седьмой день равна выручке за всю неделю, то есть последнее число — это итог недели.
Мы это проверили в коде: оно совпало с `np.sum(week1)`.

**Когда превысили половину?**
Половина недельной выручки — 12 880,5 тг. После 3-го дня накоплено 9 930, а после 4-го — 14 393, значит
порог перешли на **4-й день**. Это нормально: к середине недели уже собрано больше половины, потому что 4–6 дни
у нас самые сильные.
