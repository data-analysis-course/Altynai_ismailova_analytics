# Блок 5 — Изменение формы и объединение данных

В конце этого же файла — Блок 6 (бизнес-анализ), потому что ему нужен массив `all_sales`, который собирается здесь.

```python
import numpy as np

# Данные те же, что и в блоке 1: seed фиксированный, поэтому числа совпадают
rng = np.random.default_rng(seed=2026)

# Продажи за неделю 1: 7 дней, 3 категории товаров
week1 = rng.integers(low=500, high=2000, size=(7, 3))

# Продажи за неделю 2: 7 дней, 3 категории товаров
week2 = rng.integers(low=500, high=2000, size=(7, 3))

# Категории товаров (столбцы)
categories = np.array(["Electronics", "Clothing", "Home"])
```

## Задание 5.1 — Изменение формы `week1`

```python
# 1-2. Превращаем таблицу 7x3 в плоский одномерный массив
week1_flat = week1.reshape(-1)  # то же, что week1.flatten(), но без копии, пока это возможно
print("Одномерный массив:")
print(week1_flat)

# 3. Количество элементов
print("Элементов:", week1_flat.size, "| форма:", week1_flat.shape)

# 4. Возвращаем обратно в (7, 3)
week1_restored = week1_flat.reshape(7, 3)
print("\nВосстановленная таблица:")
print(week1_restored)
print("Совпадает с исходной:", np.array_equal(week1, week1_restored))
```

**Почему число элементов не изменилось?**
`reshape` только меняет, как мы *смотрим* на те же числа (раскладываем их по строкам и столбцам),
но сами значения не добавляет и не удаляет. Было 7 × 3 = 21 число — и осталось 21.

## Задание 5.2 — Использование `-1` в reshape

```python
# 1. Строк ровно 7, а столбцов пусть NumPy досчитает сам
reshaped_7_rows = week1.reshape(7, -1)

# 2. Столбцов ровно 3, строки считаем автоматически
reshaped_3_columns = week1.reshape(-1, 3)

# 3-4. Выводим матрицы и их формы
print("reshape(7, -1):")
print(reshaped_7_rows)
print("форма:", reshaped_7_rows.shape)

print("\nreshape(-1, 3):")
print(reshaped_3_columns)
print("форма:", reshaped_3_columns.shape)
```

**Как NumPy считает пропущенное измерение?**
Всего элементов 21. Если задано `(7, -1)`, то NumPy делит 21 на 7 и получает 3 столбца.
Если `(-1, 3)` — делит 21 на 3 и получает 7 строк. В обоих случаях получается `(7, 3)`.
Удобно, когда мы знаем только одно измерение, а второе не хочется считать вручную.

## Задание 5.3 — Когда reshape не получается

```python
demo = np.arange(1, 21)  # числа от 1 до 20, всего 20 элементов
print("demo:", demo, "| элементов:", demo.size)

# 1-3. Размеры, которые подходят: 4*5, 5*4 и 2*10 дают ровно 20
for rows, columns in [(4, 5), (5, 4), (2, 10)]:
    print(f"\nreshape({rows}, {columns}):")
    print(demo.reshape(rows, columns))

# 4. (3, 7) = 21 элемент, а у нас 20 — NumPy выдаст ошибку, ловим её, чтобы ноутбук не остановился
print("\nПробуем reshape(3, 7):")
try:
    demo.reshape(3, 7)
except ValueError as error:
    print("Ошибка:", error)
```

**Почему одни reshape работают, а другие нет?**
Правило: строки × столбцы должно равняться количеству элементов. 4 × 5 = 20, 5 × 4 = 20, 2 × 10 = 20 — подходит.
А 3 × 7 = 21 ≠ 20, один элемент пришлось бы придумать, но NumPy так не делает и выдаёт `ValueError`.
Заметьте, что при этом 20 разложить можно разными способами, пока произведение остаётся двадцатью.

## Задание 5.4 — Объединение двух недель

```python
# 1-2. vstack ставит week2 под week1 (по вертикали)
all_sales = np.vstack([week1, week2])

# 3. Показываем результат
print("all_sales:")
print(all_sales)

# 4. Форма новой таблицы
print("\nФорма:", all_sales.shape)
```

**Почему 14 строк?**
Каждая неделя — это 7 строк (дней). `vstack` складывает таблицы друг под другом: 7 + 7 = 14 дней.
Столбцов по-прежнему 3, так как категории те же. Для сравнения, `hstack` склеил бы таблицы рядом
и дал бы форму `(7, 6)`.

# Блок 6 — Бизнес-анализ

Дальше везде работаем с `all_sales` (14 дней × 3 категории).

## Задание 6.1 — Анализ категорий товаров

$$
\text{Average Sales} = \frac{\text{Total Sales}}{\text{Number of Days}}
\qquad
\text{Share of Category} = \frac{\text{Category Sales}}{\text{Total Sales}} \times 100
$$

```python
# 1. Выручка каждой категории за две недели (суммируем по дням — axis=0)
category_totals = np.sum(all_sales, axis=0)

# 2. Средняя выручка категории за день
total_number_of_days = all_sales.shape[0]
category_averages = category_totals / total_number_of_days

# 3. Доля каждой категории в общей выручке, в процентах
overall_revenue = category_totals.sum()
category_shares = category_totals / overall_revenue * 100

for index, category_name in enumerate(categories):
    print(f"{category_name:12} всего: {category_totals[index]:6}  "
          f"в среднем за день: {category_averages[index]:8.1f}  "
          f"доля: {category_shares[index]:5.1f}%")

print("\nСумма долей:", round(category_shares.sum(), 2), "%")

# 4. Категория с наибольшей выручкой
best_category = categories[np.argmax(category_totals)]
print("Лидер по выручке:", best_category)
```

**Вывод для менеджера.** Лидирует `Clothing`: 18 553 тг, примерно 35% всей выручки.
`Home` рядом (17 753 тг, около 34%), `Electronics` чуть отстаёт (16 570 тг, около 31%).
Разброс небольшой, категории вносят примерно равный вклад, зато одежда чуть лучше остальных, и именно её
стоит поддерживать в первую очередь, а для `Electronics` стоит подумать об акциях.

## Задание 6.2 — Лучшие и худшие дни

$$
\text{Daily Total Sales} = \text{Electronics} + \text{Clothing} + \text{Home}
$$

```python
# 1. Выручка за каждый день (суммируем три категории — axis=1)
daily_totals = np.sum(all_sales, axis=1)
print("Выручка по дням (14 дней):")
print(daily_totals)

# 2-3. Индексы лучшего и худшего дней; +1, потому что дни считаем с единицы
best_day_index = np.argmax(daily_totals)
worst_day_index = np.argmin(daily_totals)

# 4. Выводим дни и выручку
print(f"\nЛучший день: №{best_day_index + 1}, выручка {daily_totals[best_day_index]} тг")
print(f"Худший день: №{worst_day_index + 1}, выручка {daily_totals[worst_day_index]} тг")
```

**Как использовать в бизнесе?**
Лучший день (8-й, это первый день второй недели) стоит разобрать: была ли акция, рассылка, какой-то товар-хит —
и повторить это. Худший день (7-й, последний день первой недели) покажет, когда магазин «проседает»:
в такие дни можно запускать скидки или рекламу, а также планировать персонал и склад.

## Задание 6.3 — Сравнение недель

$$
\text{Weekly Difference} = \text{Week 2 Total} - \text{Week 1 Total}
\qquad
\text{Change \%} = \frac{\text{Week 2 Total} - \text{Week 1 Total}}{\text{Week 1 Total}} \times 100
$$

```python
# 1. Итоги по неделям (берём куски общей таблицы: первые 7 строк и последние 7)
week1_total = np.sum(all_sales[:7])
week2_total = np.sum(all_sales[7:])
print("Выручка недели 1:", week1_total)
print("Выручка недели 2:", week2_total)

# 2. Средняя дневная выручка по каждой неделе
week1_average = week1_total / 7
week2_average = week2_total / 7
print("Средняя за день, неделя 1:", round(week1_average, 1))
print("Средняя за день, неделя 2:", round(week2_average, 1))

# 3. Разница между неделями
weekly_difference = week2_total - week1_total
print("Разница (неделя 2 - неделя 1):", weekly_difference)

# 4. Изменение в процентах относительно первой недели
change_percent = weekly_difference / week1_total * 100
print(f"Изменение: {change_percent:.2f}%")

# 5. Какая неделя успешнее
more_successful_week = 2 if week2_total > week1_total else 1
print("Успешнее неделя №", more_successful_week)
```

Вторая неделя выручила на 1 354 тг больше (рост примерно 5,3%), так что **неделя 2 успешнее**.
Рост небольшой, но положительный.

## Задание 6.4 — Итоговый анализ

$$
\text{Days Below Average} = \text{Days where Daily Sales} < \text{Average Daily Sales}
$$

```python
# 1. Самая продаваемая категория
print("Лучшая категория:", categories[np.argmax(category_totals)])

# 2. Лучший день продаж
print("Лучший день: №", np.argmax(daily_totals) + 1)

# 3. Неделя с более высокой средней выручкой
print("Неделя с большей средней:", 2 if week2_average > week1_average else 1)

# 4. Дни, когда выручка была ниже средней по всем 14 дням
average_daily_sales = np.mean(daily_totals)
below_average_mask = daily_totals < average_daily_sales
days_below_average = np.where(below_average_mask)[0] + 1  # номера дней с единицы

print(f"\nСредняя дневная выручка: {average_daily_sales:.1f} тг")
print("Дни ниже средней:", days_below_average)
print("Их выручка:", daily_totals[below_average_mask])
print("Всего таких дней:", below_average_mask.sum(), "из", daily_totals.size)
```

**Краткий вывод для менеджера.**
- За две недели магазин заработал 52 876 тг, в среднем около 3 777 тг в день.
- Лучшая категория — `Clothing`, но разница между категориями небольшая.
- Вторая неделя лучше первой примерно на 5%.
- Самый сильный день — 8-й (5 420 тг), самый слабый — 7-й (2 342 тг). Дни 1, 2, 3, 7, 10, 12, 13 оказались ниже
  среднего, то есть примерно половина периода работает ниже нормы. Стоит присмотреться к ним и поддержать продажи
  скидками или рекламой, а удачные дни разобрать и повторить.

## Ответы на контрольные вопросы

1. **Зачем NumPy в аналитике?** Быстро считает по большим таблицам без циклов: фильтры, суммы, преобразования.
2. **`axis=0` и `axis=1`?** `axis=0` — операция вдоль строк, результат по каждому столбцу; `axis=1` — вдоль столбцов, результат по каждой строке.
3. **Чем булева фильтрация отличается от обычной индексации?** Обычный индекс выбирает по позиции, а булева маска — по условию на значения.
4. **Зачем накопленная сумма?** Показывает, как выручка нарастает со временем и когда достигается нужная доля или план.
5. **`vstack` и `hstack`?** `vstack` добавляет строки (друг под другом), `hstack` — столбцы (рядом).
6. **Чем векторизация удобнее циклов?** Короче, понятнее и намного быстрее.
7. **Чем полезен broadcasting?** Позволяет применить к столбцам или строкам разные коэффициенты (скидки, рост) одной операцией.
8. **Как reshape помогает готовить данные?** Приводит данные к нужной форме для расчётов и объединения, не меняя сами значения.
