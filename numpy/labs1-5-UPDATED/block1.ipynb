Работаем с продажами интернет-магазина за две недели. Выручка указана в тенге.

## Задание 1.1 — Генерация данных о продажах

```python
import numpy as np

# Генератор случайных чисел с фиксированным seed, чтобы результат повторялся
rng = np.random.default_rng(seed=2026)

# Продажи за неделю 1: 7 дней (строки), 3 категории товаров (столбцы)
week1 = rng.integers(low=500, high=2000, size=(7, 3))

# Продажи за неделю 2: такая же форма, но уже другие числа
week2 = rng.integers(low=500, high=2000, size=(7, 3))

# Названия категорий — они соответствуют столбцам
categories = np.array(["Electronics", "Clothing", "Home"])

print("Продажи за неделю 1:")
print(week1)

print("\nПродажи за неделю 2:")
print(week2)

print("\nКатегории:")
print(categories)
```

```python
# Проверяем характеристики обоих массивов
for name, sales_table in [("week1", week1), ("week2", week2)]:
    print(f"--- {name} ---")
    print("shape (форма):          ", sales_table.shape)
    print("ndim (число измерений): ", sales_table.ndim)
    print("size (всего элементов): ", sales_table.size)
    print("dtype (тип данных):     ", sales_table.dtype)
    print()
```

Что получилось: у каждой недели форма `(7, 3)`, то есть 2 измерения и 7 × 3 = 21 элемент.
Тип данных — целые числа (`int64`), что логично для выручки в тенге.

## Задание 1.2 — Разбираемся в данных

**Что означает каждая строка?**
Одну продажную дату, то есть один день недели. В `week1` семь строк — семь дней первой недели.

**Что означает каждый столбец?**
Одну категорию товаров: `Electronics`, `Clothing` или `Home`. В ячейке лежит выручка этой категории за конкретный день.

**Зачем фиксировать `seed`?**
Данные случайные, и без `seed` при каждом запуске получались бы новые числа. Тогда мои ответы
(например, «лучший день — 8-й») не совпадали бы с тем, что увидит преподаватель, и результаты нельзя было бы проверить.
С `seed=2026` генератор выдаёт одну и ту же последовательность, так что анализ воспроизводим.
