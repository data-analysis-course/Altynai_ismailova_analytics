# Блок 2 — Выбор и фильтрация данных

```python
import numpy as np

# Данные те же, что и в блоке 1: seed фиксированный, поэтому числа совпадают
rng = np.random.default_rng(seed=2026)

# Продажи за неделю 1: 7 дней, 3 категории товаров
week1 = rng.integers(low=500, high=2000, size=(7, 3))

# Продажи за неделю 2: 7 дней, 3 категории товаров
week2 = rng.integers(low=500, high=2000, size=(7, 3))

# Категории товаров (столбцы)
categories = np.array(["Electronics", "Clothing", "Home"])
```

```python
print("week1:")
print(week1)
```

## Задание 2.1 — Индексы и срезы

Напоминаю: в `week1[строка, столбец]` нумерация идёт с нуля. Столбцы: 0 — Electronics, 1 — Clothing, 2 — Home.

```python
# 1. Продажи за первый день (вся первая строка)
first_day_sales = week1[0]
print("Первый день:", first_day_sales)

# 2. Продажи за последний день (индекс -1 — это конец массива)
last_day_sales = week1[-1]
print("Последний день:", last_day_sales)

# 3. Electronics за всю неделю: все строки, нулевой столбец
electronics_all_week = week1[:, 0]
print("Electronics за неделю:", electronics_all_week)

# 4. Clothing за первые 3 дня: строки 0..2, столбец 1
clothing_first_three_days = week1[:3, 1]
print("Clothing, первые 3 дня:", clothing_first_three_days)

# 5. Дни с 3-го по 5-й включительно.
# Дни считаем с единицы, поэтому индексы 2, 3, 4 — конец среза 5 не включается
days_three_to_five = week1[2:5]
print("Дни с 3 по 5:")
print(days_three_to_five)
```

## Задание 2.2 — Булева фильтрация

```python
# 1. Все значения выручки больше 1500
above_1500 = week1[week1 > 1500]
print("Больше 1500:", above_1500)

# 2. Все значения меньше 800
below_800 = week1[week1 < 800]
print("Меньше 800:", below_800)

# 3. Значения от 1000 до 1500 включительно — два условия через &, каждое в скобках
between_1000_and_1500 = week1[(week1 >= 1000) & (week1 <= 1500)]
print("От 1000 до 1500:", between_1000_and_1500)

# 4. Сколько дней Electronics продавалась больше чем на 1200
electronics_sales = week1[:, 0]
days_electronics_above_1200 = np.sum(electronics_sales > 1200)
print("Дней с Electronics > 1200:", days_electronics_above_1200)
```

**Какие условия использовал.**
- `week1 > 1500` — выбирает «хорошую» выручку;
- `week1 < 800` — «слабую»;
- `(week1 >= 1000) & (week1 <= 1500)` — диапазон, оба края включены, поэтому `>=` и `<=`;
- `electronics_sales > 1200` — условие даёт массив из `True`/`False`, а `np.sum` считает `True` как единицы, и получается количество дней.

**Почему `&` и `|`, а не `and` и `or`?**
`and` / `or` в Python работают с одним значением «правда/ложь» и не умеют сравнивать массивы поэлементно
(выпадет ошибка про неоднозначность истинности массива). `&` и `|` применяются к каждому элементу отдельно.
Ещё важно: каждое сравнение берём в скобки, потому что у `&` приоритет выше, чем у `>` и `<`.

**Какую бизнес-задачу решает булева фильтрация?**
Например, находить дни и товары с очень высокой или очень низкой выручкой: понять, где были всплески, а где провалы,
выделить аномалии, отобрать крупные заказы или слабые категории для акции.
