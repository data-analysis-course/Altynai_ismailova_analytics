# Блок 4 — Преобразование продаж

```python
import numpy as np

# Данные те же, что и в блоке 1: seed фиксированный, поэтому числа совпадают
rng = np.random.default_rng(seed=2026)

# Продажи за неделю 1: 7 дней, 3 категории товаров
week1 = rng.integers(low=500, high=2000, size=(7, 3))

# Продажи за неделю 2: 7 дней, 3 категории товаров
week2 = rng.integers(low=500, high=2000, size=(7, 3))

# Категории товаров (столбцы)
categories = np.array(["Electronics", "Clothing", "Home"])
```

## Задание 4.1 — Векторизованное повышение цен

$$
\text{New Value} = \text{Old Value} \times 1.12
$$

```python
# 1. Копия, чтобы не испортить исходный week1
week1_copy = week1.copy()

# 2. Увеличиваем сразу все значения на 12% — без единого цикла
week1_increased = week1_copy * 1.12

# 3. Показываем исходный и новый массивы
print("Исходный week1:")
print(week1)
print("\nПосле повышения на 12% (дробные числа):")
print(week1_increased)

# 4. Округляем до целых тенге
week1_increased_rounded = np.round(week1_increased).astype(int)
print("\nПосле округления:")
print(week1_increased_rounded)

# Проверим, что исходный массив остался нетронутым
print("\nweek1 не изменился:", np.array_equal(week1, week1_copy))
```

**Почему не нужен цикл `for`?**
Умножение массива на число в NumPy — векторизованная операция: библиотека сама применяет её ко всем элементам сразу,
причём внутри на быстром C-коде. Код короче и читается проще (одна строка вместо двух вложенных циклов),
а на больших таблицах работает на порядки быстрее обычного питоновского цикла.

## Задание 4.2 — Категоризация продаж

```python
# 1. Булева маска: True там, где выручка больше 1500
high_sales_mask = week1 > 1500
print("Маска (True = больше 1500):")
print(high_sales_mask)

# 2. np.where: если условие True — берём "High", иначе "Low"
sales_labels = np.where(week1 > 1500, "High", "Low")

# 3. Оригинальный массив
print("\nОригинал:")
print(week1)

# 4. Тот же массив, но с метками
print("\nМетки:")
print(sales_labels)

# Сколько ячеек получили каждую метку
print("\nHigh:", np.sum(sales_labels == "High"), "| Low:", np.sum(sales_labels == "Low"))
```

**Как работает `np.where`?**
Это «векторный if-else»: `np.where(условие, значение_если_True, значение_если_False)`.
NumPy проверяет условие для каждой ячейки и подставляет либо первое, либо второе значение.
Форма результата та же, что у условия, здесь `(7, 3)`. Граничный случай: ровно 1500 попадает в `"Low"`,
потому что условие строгое (`>`), как и требует задание.
