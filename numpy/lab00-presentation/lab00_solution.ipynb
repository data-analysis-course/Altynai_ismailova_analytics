## Lab: Computing Monthly Averages


```python
# Lab: Computing Monthly Averages

# Parameters 
days = 360
products_count = 4


# Simulate daily sales units
daily_sales = np.random.randint(low=10, high=101, size=(days, products_count))

# Reshape into (months, days_per_month, products)
monthly_sales = daily_sales.reshape(12, 30, 4)

# Calculate average daily sales per product per month
monthly_avg_sales = monthly_sales.mean(axis=1)
# print(monthly_sales) 
print(monthly_avg_sales) 


# result output :
#  [[56.56666667 51.9        49.96666667 56.2       ]
#  [58.36666667 49.96666667 60.06666667 53.8       ]
#  [48.93333333 55.66666667 60.26666667 53.93333333]
#  [52.13333333 50.         51.26666667 56.9       ]
#  [54.26666667 50.03333333 63.13333333 50.66666667]
#  [60.2        57.53333333 51.63333333 56.3       ]
#  [52.6        50.03333333 53.53333333 52.66666667]
#  [56.66666667 47.26666667 47.73333333 56.13333333]
#  [52.46666667 57.96666667 54.53333333 51.9       ]
#  [53.1        42.53333333 56.2        49.3       ]
#  [53.7        44.83333333 50.5        56.26666667]
#  [62.76666667 58.76666667 53.66666667 61.3       ]]
```

# Lab: Performance Comparison

```python
# Lab: Performance Comparison

# python loop - inefficient
def sum_with_loop(data): 
    total = 0
    for row in data:
        for item in row: 
            total += item
    return total
# Timing
%timeit sum_with_loop(daily_sales)

# Using NumPy's vectorized sum - very fast and efficient
%timeit np.sum(daily_sales)

# result:
# 210 μs ± 4.05 μs per loop (mean ± std. dev. of 7 runs, 1,000 loops each)
# 1.17 μs ± 17.4 ns per loop (mean ± std. dev. of 7 runs, 1,000,000 loops each)
# so numpy is 180 times faster
```

# Lab: Weekly Sales Aggregation

```python
# Lab: Weekly Sales Aggregation

days = 364
products_count=4
daily_sales = np.random.randint(low=1, high=101,size=(days,products_count))
# Reshape into (weeks, days_per_week, products) 
weekly_sales_data = daily_sales.reshape(52, 7, 4)
# 
# Sum across 'days' axis to get weekly totals 
sum_of_weekly_sales = weekly_sales_data.sum(axis=1)

print("Shape of sum_of_weekly_sales:", sum_of_weekly_sales.shape) 
print("Total sales for the first week:")
print(sum_of_weekly_sales[0])

# output:
# Shape of sum_of_weekly_sales: (52, 4)
# Total sales for the first week:
# [407 314 397 420]
```
