```python
a=[1,2,3]
b=[8,15,23]
plt.plot(a,b, color="red", marker="s", linestyle='--' , label="my love count")
plt.title("how i love isken")
plt.xlabel("my love proportion")
plt.ylabel("time")
plt.legend()
plt.show()
```
