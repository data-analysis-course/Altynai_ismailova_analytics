# Part 1. Theory: Working with CSV Files

CSV (Comma-Separated Values) is the most common format for storing tabular data. Each row is one observation, columns are separated by commas.

## 1.1 Loading a CSV file

The main function is `pd.read_csv()`.

```python
df = pd.read_csv("ecommerce_orders.csv")
```

Key parameters:

- `sep` — separator (default `","`)
- `encoding` — file encoding (default `"utf-8"`; use `"cp1251"` for some Windows/Excel files)
- `parse_dates` — list of columns to convert to datetime
- `nrows` — read only first N rows (useful for large files)
- `usecols` — load only selected columns


````python
import pandas as pd

# Basic load
df = pd.read_csv("ecommerce_orders.csv")

# Check the result
df.head(len(df)) # Display all rows using head()

````

## 1.2 Checking the data after loading

Always check: shape, column types, missing values, first rows.


````python
print(df.shape)          # (rows, columns)
print(df.dtypes)         # column types
print(df.isna().sum())   # missing values per column
df.info()
````

## 1.3 Converting dates

`Order Date` is loaded as text (object). For time-series analysis it must be converted to datetime.


````python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df.dtypes
````

## 1.4 Saving (exporting) to CSV

Use `df.to_csv()`.

Key parameters:

- `index=False` — do not save the row index (almost always needed)
- `sep` — separator
- `encoding` — `"utf-8"` is standard; `"utf-8-sig"` opens correctly in Excel


````python
# Save a copy
df.to_csv("ecommerce_orders_copy.csv", index=False, encoding="utf-8-sig")
print("Saved!")
````

## Exercise 3. Crosstabs & Pivot Tables

Create a **crosstab** showing the number of orders by Region and Category.  
Then create a **pivot table** showing the **total revenue** by these dimensions.

```python
pd.crosstab(df["Region"], df["Category"])
pd.pivot_table(df, values="Revenue", index="Region", columns="Category", aggfunc="sum")
```

*Tip: use `fill_value=0` to handle missing combinations in pivot tables.*


````python
# Your solution here
cross_tab = pd.crosstab(df["Region"], df["Category"])
pivot_table = pd.pivot_table(df, values="Revenue", index="Region", columns="Category", aggfunc="sum", fill_value=0)
# fill value pretends that n/a values are 0
print(cross_tab)
print("\n pivot table:")
print(pivot_table)
````

## Exercise 4. Ranking & Sorting

Identify the **top 3 products by revenue for each region** using ranking techniques.

```python
df["RevenueRank"] = df.groupby("Region")["Revenue"].rank(method="dense", ascending=False)
top3_per_region = df[df["RevenueRank"] <= 3]
```

*Tip: use `method="dense"` for consecutive integer ranks.*


````python
# Your solution here
df["RevenueRank"] = df.groupby("Region")["Revenue"].rank(method="dense", ascending=False)
top3_by_region = df[df["RevenueRank"] <= 3].sort_values(["Region", "RevenueRank"])

print(top3_by_region[["Region", "Product", "Revenue", "RevenueRank"]])

# dense means that if we have the same value like 500, they bith count as 1st place
````

## Exercise 6. Data Reshaping

Use `melt()` to transform the e-commerce data from **wide to long** format, then use `pivot()` to transform it back to **wide** format.

```python
# Wide to Long
df.melt(id_vars=["Region"], value_vars=["Revenue"])

# Long to Wide
df.pivot(index="Region", columns="Category", values="Revenue")
```

*Tip: `melt()` converts columns into rows; `pivot()` does the opposite.*


````python
# Your solution here
# Wide to Long
long_df = df.melt(id_vars=["Region"], value_vars=["Revenue"])
print(long_df.head())

# Long to Wide
# pivot() needs one row per Region/Category pair, so sum the orders first
wide_df = (df.groupby(["Region", "Category"])["Revenue"].sum().reset_index().pivot(index="Region", columns="Category", values="Revenue"))
print(wide_df)
````
