### Experiment with `dropna()`
Done on a fresh copy of the data **before** the salaries were filled, so the effect is visible.

```python
test = employees.copy()
test["Salary"] = pd.to_numeric(test["Salary"], errors="coerce")
test.loc[0, "Age"] = np.nan      # one more NaN, in a different column, to show subset/thresh
print("original rows:", len(test))

print("default (axis=0, how='any'):      ", len(test.dropna()))
print("how='all':                        ", len(test.dropna(how="all")))
print("subset=['Salary']:                ", len(test.dropna(subset=["Salary"])))
print("subset=['Salary','Age']:          ", len(test.dropna(subset=["Salary", "Age"])))
print("thresh=9 (keep rows with >=9 non-NaN):", len(test.dropna(thresh=9)))
print("thresh=8:                         ", len(test.dropna(thresh=8)))
print("axis=1 (drop columns with any NaN):", list(test.dropna(axis=1).columns))
```

### Task 2 — explanation
* `"unknown"` is a *string*, so pandas does not see it as missing. `replace("unknown", np.nan)` makes it a real missing value, and `pd.to_numeric(..., errors="coerce")` converts the column to `float64` (anything unparsable would also become `NaN`).
* After this, `Salary` has 3 missing values (3 / 15 = **20 %**). They were filled with the **median** (not the mean) because the median is robust to outliers and keeps the distribution realistic.
* What each `dropna()` parameter changes:
  * `axis=0` (default) drops **rows**; `axis=1` drops **columns** that contain NaN.
  * `how="any"` (default) drops a row if *any* value is missing; `how="all"` only if *all* values are missing (nothing is dropped here).
  * `subset=[...]` only looks at the listed columns — dropping rows with missing `Salary` removes 3 rows, adding `Age` removes one more.
  * `thresh=n` keeps a row only if it has at least *n* non-missing values, which is a softer rule than `how="any"`.
* Dropping data is costly with only 15 rows, which is why imputation was chosen for `Salary`.

## Task 3 — Standardize Department Names

```python
print("Before:", sorted(df["Department"].unique().tolist()))

df["Department"] = (
    df["Department"]
    .str.strip()
    .str.replace("Information Tecnology", "IT", regex=False)
    .str.replace(r"^Financ\w*$", "Finance", regex=True, case=False)   # Financing / finance / Finance
    .str.upper()
    .replace({"FINANCE": "Finance"})
)

print("After: ", sorted(df["Department"].unique().tolist()))
df["Department"].value_counts()
```

```python
# the same kind of tidy-up for the other text columns
df["LastName"] = df["LastName"].str.strip().str.title()
df["Gender"] = df["Gender"].str.strip().str.upper()
print(df["Gender"].unique(), df["LastName"].tolist())
```

```python
# incorrect age
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
print(df.loc[(df["Age"] < 18) | (df["Age"] > 70), ["EmployeeID", "FirstName", "LastName", "Age"]])
```

```python
# decision: an age of 150 is impossible -> treat it as missing and fill it with the median age
df.loc[(df["Age"] < 18) | (df["Age"] > 70), "Age"] = np.nan
median_age = int(df["Age"].median())   # 30.5 -> 30, so Age can become an integer column
df["Age"] = df["Age"].fillna(median_age)
print("median age used:", median_age)
df["Age"].describe()
```

### Task 3 — explanation
* `str.strip()` removes the surrounding spaces (`" IT "`), `str.replace()` fixes the misspellings (`Information Tecnology` → `IT`, `Financing`/`finance` → `Finance`) and `str.upper()` makes the case consistent (`it` → `IT`). The unique values are now only **IT, HR, Finance**, matching `departments["DepartmentName"]`.
* `Age` 150 (employee 105, Nurlan Abdyldaev) is a data-entry error — nobody is 150 and we cannot know the real value. Deleting the whole employee would lose a valid salary and hire date, so the age was set to `NaN` and replaced with the **median age** (30.5, rounded down to 30 so the column can stay an integer). Other reasonable choices: leave it `NaN` (kept as nullable `Int64`) or ask HR for the real value.

```python
# object vs category: memory usage
dept_obj = df["Department"].astype("object")
print("object  :", dept_obj.memory_usage(deep=True), "bytes")
print("category:", df["Department"].memory_usage(deep=True), "bytes")
```

### Task 5 — explanation
* `Int64` (capital I) is a **nullable integer**: unlike NumPy `int64` it can store `<NA>`, so a column of ages can stay integer even if a value is missing.
* **object vs category:** `object` stores every value as a separate Python string. `category` stores each unique label once and keeps small integer codes per row, which uses less memory, is faster for `groupby`/comparisons, and documents that the column has a fixed set of values (`IT`, `HR`, `Finance`). Use `object` for free text (names), `category` for low-cardinality labels.
* `PerformanceScore` held the text `"4.0"` (employee 114), so it was converted with `pd.to_numeric` to `float64`; no value was lost.

## Task 6 — Parse Hire Dates

```python
df["HireDate"] = pd.to_datetime(df["HireDate"], format="%Y-%m-%d", errors="coerce")
print(df["HireDate"].dtype)
print("unparsed dates:", df["HireDate"].isna().sum())
```

```python
df["HireYear"] = df["HireDate"].dt.year
df["HireMonth"] = df["HireDate"].dt.month
df["HireWeekday"] = df["HireDate"].dt.day_name()

# years worked up to a fixed reference date, so the result is reproducible
reference_date = pd.Timestamp("2025-01-01")
df["YearsWorked"] = ((reference_date - df["HireDate"]).dt.days / 365.25).round(1)

df[["EmployeeID", "HireDate", "HireYear", "HireMonth", "HireWeekday", "YearsWorked"]]
```

### Task 6 — explanation
* `pd.to_datetime` turns the text into `datetime64[ns]`, which allows date arithmetic and the `.dt` accessor. `errors="coerce"` would turn any unparsable date into `NaT` instead of raising an error (none were found).
* `.dt.year`, `.dt.month` and `.dt.day_name()` extract the parts used for time-based analysis (e.g. hires per year).
* **Years worked** = (reference date − hire date) in days ÷ 365.25. A fixed reference date (`2025-01-01`) is used so the notebook gives the same result every time; use `pd.Timestamp.today()` instead for "up to today".

## Task 7 — Rename Columns and Merge Datasets

```python
df = df.rename(columns={"Salary": "MonthlySalary", "HireDate": "EmploymentStartDate"})
print(df.columns.tolist())
```

```python
# Department is a category column; cast to object so it can be matched with a plain-text key
df["Department"] = df["Department"].astype("object")

merged = df.merge(
    departments,
    how="left",
    left_on="Department",
    right_on="DepartmentName",
)
print("rows before merge:", len(df), "| after merge:", len(merged))
print("missing values created by the merge:")
print(merged[["DepartmentID", "DepartmentName", "Location"]].isna().sum())
merged[["EmployeeID", "FirstName", "Department", "DepartmentID", "Location"]]
```

### Task 7 — explanation
* `rename(columns={...})` takes an `old: new` mapping; names not listed are left unchanged.
* The merge key is `Department` (employees) = `DepartmentName` (departments), so `left_on`/`right_on` are used. `how="left"` keeps **all** employees, adding `DepartmentID` and `Location`.
* Because the department names were standardised in Task 3, **no missing values** appeared and the row count stayed the same (no new duplicates either). Without the cleaning, `" IT "`, `it`, `Financing` and `Information Tecnology` would not have matched and would have produced `NaN` in `DepartmentID` and `Location`.

**Methods and parameters used**

**Result:** 14 unique employees, no missing values, no duplicates, correct data types, consistent department names, and a clean join with the department table (location added for everyone).

**Why data preparation matters before HR analysis:** averages and counts computed on dirty data are wrong — a duplicate inflates headcount, `"unknown"` or `NaN` breaks salary statistics, an age of 150 distorts the mean age, and inconsistent department names split one department into several groups or break joins. Clean, correctly-typed data makes every later result (salary by department, tenure, performance) trustworthy.
